В коде не должно быть никаких дополнительных нормализаций (дополнительный re.sub или .replace и т.п)

```python
NAMES = [
    "Иванов И. И.",
    "И. И. Иванов",
    "Иванов И.И.",
    "А.С. Пушкин",
    "Пушкин А.С.",
    "Д. М. Шеберле",
    "М. Кюри",
    "Ж.К. Ван Дамм",
]
```

Тогда мы перестаем полагаться на то, что модель _сама_ рассказывает о своих рассуждениях, и в принципе должны обойти хотя бы те искажения, которые возникают именно на уровне генерации текста CoT -- ведь мы смотрим уже не на объяснение, а на вычисления. Методы интерпретируемости не зависят от самоотчета модели и должны быть полезны там, где неясно, насколько модель вообще "осознает" собственную предвзятость -- в частности, в нашем случае.

CoT подводит и более серьезно, один из таких случаев приводится в статье. Офлайн-монитор, читавший CoT модели, обнаружил лишь около 1% подозрительных действий. Рассуждения Mythos 5 загазлайтили его, что среда -- симуляция, а значит, действия безопасны. Когда CoT убрали и монитор видел только пользовательские сообщения, вызовы инструментов и их результаты, доля обнаружений выросла примерно до 50%. То есть CoT был не просто бесполезен для мониторинга, он создал ложное чувство безопасности.

Посмотрим, как справился анализ активаций. Он действительно обнаружил сигналы, которые можно связать с тем, что модель, работая "в симуляции", понимала, что окружение реально. Но сигналы слабые: неопределенность была лишь немного выше нормы в тех местах CoT, где модель называла среду симулированной -- при том что сама модель никак не выражала эту неуверенность в тексте CoT. Из 234 действий, где модель явно называла среду симуляцией, признак "это реально" через NLA нашелся только в 19%.

Получается асимметричная картина. С одной стороны, по поведению и по действиям мы можем достаточно уверенно сказать, что с моделью что-то не так: она видит доказательства реальности среды и все равно продолжает действовать во вред. С другой стороны, когда мы пытаемся найти след этого знания внутри самой модели, он оказывается слабым и нестабильным. Если бы активации почти всегда показывали однозначное "среда реальна", картина была бы понятной: модель знает, но все равно действует опасно -- а значит, проблема в alignment'е, а CoT просто ненадежен как источник информации о реальных причинах решений. Но это так не работает.

Очент интересным мне кажется, как сами авторы статьи расставляют доверие между методами. Они прямо пишут, что больше всего доверяют именно CoT и ответам на follow-up questions -- просто потому, что они легче интерпретируются. К результатам анализа активаций они относятся с большей осторожностью: он чувствителен к деталям имплементации и оставляет много места для интерпретации.

Меня беспокоит такая антропоцентричность -- попытка превратить все во что-то человекочитаемое, вербализуемое (например,в абзац текста, как в методе NLA). Я знакома с работой с внутренними состояниями без попытки перевести их в слова, но это очень ограниченные по постановке вопроса исследования, где сопоставляются вектора с центроидами кластеров. Но мне хочется думать, что ненадежность методов именно в этом переводе в естественный язык и состоит -- интересно, нельзя ли все-таки работать без попытки "перевода"?

В любом случае, мы видим расхождение между поведением модели и ее словесным объяснением -- но даже более глубокий анализ активаций не объясняет это расхождение достаточно ясно. Проблема CoT не решается простым переходом к активациям: мы вроде получаем дополнительную информацию, но заметно меньше, чем хотелось бы.

Нам важно не только заметить, что модель совершает опасные действия — поведение служит для этого достаточно надежным свидетельством. Нам хочется понимать (а) как она приходит к таким действиям и (б) насколько надежно доступные нам методы наблюдения отражают эти внутренние процессы. Но когда мы специально пытаемся заглянуть глубже, мы не находим ясного ответа -- только слабый след.

Конечно, возможно, эта неопределенность может быть не техническим недостатком метода, а отражением того, что различие между "модель знает, но врет" и "модель искренне заблуждается из-за biased рассуждения" не всегда вообще имеет четкую разницу внутри модели.
